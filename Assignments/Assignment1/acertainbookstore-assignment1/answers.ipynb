\section{Provide a short description (2–4 paragraphs) of your implementation and tests:}

\subsection*{(a) How your implementation and tests enforce all-or-nothing semantics.}
Our implementation enforces all-or-nothing, by first running our loop that validates the full request to completion before running our second loop that makes all the changes. This ensures that the changes are only made if all the changes are valid.

Our tests enforce this by testing both valid and invalid requests so that the valid requests makes sure to get updated and the invalid requests make sure to not update anything to enforce all-or-nothing semantics. Specifically we test that when calling rateBooks with a valid rating and an invalid rating in the same request, none of the ratings are applied.

\subsection*{(b) How you verify that the service behaves correctly regardless of whether it is accessed via RPCs or local calls.}
We verify that the service behaves correctly regardless of whether it is accessed via RPCs or local calls by running the same test suite in two ways. Firstly we run the tests with localTest = true, which accesses the bookstore directly through local method calls. Afterwards we run the tests with localTest = false which accesses the bookstore through the HTTP proxies and RPC communication with the server. If the same tests pass in both ways this provides evidence that the service behaves consistent regardless of how it is accessed.

\section{The architecture is described as strongly modular. Explain this in this context:}

\subsection*{(a) In what sense is the architecture strongly modular? (1–2 sentences)}
The architecture is strongly modular as it seperates the system into different components that each have their own role in the system. For example the communication layer, business logic and client are all seperated into their own components, that do not rely on each others internal details. This ensures that you can write very general interfaces for example where you can both test for https and local calls in the same test implementations

\subsection*{(b) What kind of isolation and protection does the architecture provide between the two client types and the bookstore service? (1–2 sentences)}
The architecture provides logical isolation by giving the two client types separate interfaces and HTTP proxies: the BookStore client can access customer-facing operations while the StockManager client can access stock-management operations. It also protects the bookstores internal data by returning read-only or immutable copies to clients so they cannot directly modify the servers internal state.

\subsection*{(c) How does running clients and services locally in the same JVM, as in the test cases, affect modularity? (1–2 sentences)}
Running clients and services locally in the same JVM does not remove the logical modularity, since they still interact through the interfaces. However, it reduces the physical isolation between them, and when tests call CertainBookStore directly they bypass the HTTP proxies and RPC communication layer therefore this part of the architecture is not tested.

\section{We have studied three types of RPC semantics: at-least-once, at-most-once, and exactly-once. Which semantics is implemented in the architecture? Justify your answer. (1 paragraph)}
The semantics implemented in the architecture is at-most-once, as each proxy creates and sends an HTTP request once, and if any failure happens then an exception is thrown. There is also no automatic retry implemented, so a request will never happen more than once, but a request may execute zero times if it is lost.
