# Exercise 1

## What is the precedence graph for this schedule ?
T2 -> T1 (y)
T1 -> T3 (x)
T2 -> T3 (x)


## Is the schedule conflict serializable ?
Yes, the schedule is conflict serializable, as there is no cycle in the precedence graph. It can be serialized by T2, T1, T3

## Could the schedule have been generated by a scheduler using Strict 2PL ?
No? As T2 would have to let go of lock on y before it can let go of the lock on x

## Could the schedule have been generated by a scheduler using 2PL ?
Yes, T2 can acquire the lock on x and afterwards release the lock on y so T1 can get the lock

## Are there conflicts between the following pairs of transactions: {T1,T2},{T1,T3},{T2,T3} ?
{T1,T2}: Yes, as T1 writes to both x and z while T2 read both x and z. (T2 -> T1)

{T1,T3}: Yes, as T1 writes to x while T3 reads x. (T3 -> T1)

{T2,T3}: No, T3 just reads x and T2 never writes to x.

## Is the schedule conflict serializable ?
T2 -> T1 (x)
T2 -> T1 (z)
T3 -> T1 (x)

Yes, the schedule is conflict serializable, as there is no cycle in the precedence graph. It can be serialized by T2, T3, T1 or T3, T2, T1

## Could the schedule have been generated by a scheduler using Strict 2PL ?
No? As T2 would have to let go of lock on x before it can let go of the lock on z and y

## Could the schedule have been generated by a scheduler using 2PL ?
Yes, as T2 can acquire all its locks before letting go of x.


# Exercise 3

## The ACID properties are: Availability, Consistency, Integrity and Durability
False, the ACID properties are: Atomicity, Consistency, Isolation and Durability

## Strict 2PL is a concurrency control protocol that has a lock acquisition phase and a lock release phase, so Strict 2PL may suffer from cascading aborts.
False, since Strict 2PL will not suffer from cascading aborts. Cascading aborts happen when a write lock is released, and another transaction reads that value, but the original transaction fails leading to the other transaction also aborting. This will not happen in Strict 2PL as all locks are only released when the full transaction is done, and it is certain it will not abort.

## Strict 2PL may lead to deadlocks.
True, as in Strict 2PL transaction 1 might lock A, transaction 2 might lock B, and then transaction 1 might wait for a lock on B while transaction 2 also waiting for lock A causing a deadlock 


# Exercise 4

## Locking is a pessimistic approach in which conflicts are prevented.
True, since even if we did not have locks, we might still not have conflicts

## There are concurrency control protocols based on locking that prevent deadlocks altogether, and therefore there protocols do not need to implement deadlock detection.
True, Conservative Strict Two Phase Locking and Conservative Two Phase Locking both prevent deadlocks altogether.

## In optimistic concurrency control, transactions are allowed to issue writes during the Read Phase, but these writes are not made in-place directly to the database.
True, in optimistic concurrency control, transactions are allowed to issue writes during the Read Phase, but these writes are not made in-place directly to the database. Instead, they are stored in a temporary buffer and only committed if no conflicts are detected.
